```python
import requests

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/85.0.4183.121 Safari/537.36"
}
url = "https://en.wikipedia.org/wiki/List_of_largest_trading_partners_of_India"
html = requests.get(url, headers=headers).text
html
```

```python
soup = BeautifulSoup(html, "html.parser")
print(soup)
```

```python
print(soup.prettify())
```

```python
table = soup.find("table")
print(table)
```

```python
table1 = soup.find("table" , class_="wikitable sortable")
print(table1)
```

```python
all_sortable_tables = soup.find_all("table", class_="wikitable sortable")
```

```python
import pandas as pd

# The first table was 'Exports from these countries to India'
# The second 'wikitable sortable' table is 'India\'s largest trading partners'
table_with_rank_trade_balance = all_sortable_tables[1]
df_partners = pd.read_html(str(table_with_rank_trade_balance))[0]

display(df_partners.head())
```

```python
trs = table1.find_all("tr")[1:]
rows = []
for tr in trs:
  row=[td.get_text(strip=True) for td in tr.find_all(["td"])]
  rows.append(row)
rows
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(rows, columns= cols)
```
