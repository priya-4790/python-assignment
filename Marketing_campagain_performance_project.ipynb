```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
url = "https://raw.githubusercontent.com/ArchanaInsights/Datasets/main/marketing_campaign.csv"
df = pd.read_csv(url)

```

```python
print("\n--- Shape (rows, columns) ---")
print(df.shape)
```

```python
print("\n--- Info (data types + non-null) ---")
print(df.info())
```

```python
print("\n--- Descriptive Statistics ---")
print(df.describe())
```

```python
print("\n--- Unique Campaign_ID ---")
print(df['Campaign_ID'].nunique())
```

```python
print("\n--- Unique Locations ---")
print(df['Location'].unique())
```

```python
print("\n--- Unique Customer Segments ---")
print(df['Customer_Segment'].unique())
```

```python
print("\n--- Count of Campaign_Type ---")
print(df['Campaign_Type'].value_counts())
```

```python
print("\n--- Count of Channel_Used ---")
print(df['Channel_Used'].value_counts())
```

```python
print(df['Date'].head(10))
print(df['Date'].dtype)

```

```python
df['Date'] = pd.to_datetime(df['Date'], errors='coerce', dayfirst=True, infer_datetime_format=True)
```

```python
print("Failed dates (NaT):", df['Date'].isna().sum())
```

```python
df = df.dropna(subset=['Date'])
df.sort_values('Date', inplace=True)
```

```python
print("Fixed!")
print(df['Date'].head())
```

```python
# --- CAMPAIGN PERFORMANCE ---


# 1. Scatter: Acquisition_Cost vs ROI
plt.figure(figsize=(7,5))
sns.scatterplot(x='Acquisition_Cost', y='ROI', data=df, hue='Campaign_Type')
plt.title('Acquisition Cost vs ROI')
plt.show()
```

```python
# 2. Bar: Avg Conversion_Rate for Channel_Used by Campaign_Type
plt.figure(figsize=(10,5))
avg_conv = df.groupby(['Channel_Used','Campaign_Type'])['Conversion_Rate'].mean().reset_index()
sns.barplot(x='Channel_Used', y='Conversion_Rate', hue='Campaign_Type', data=avg_conv)
plt.title('Avg Conversion Rate by Channel and Campaign Type')
plt.xticks(rotation=20)
plt.show()
```

```python
# 3. Box Plot: Engagement_Score across Campaign_Type
plt.figure(figsize=(8,5))
sns.boxplot(x='Campaign_Type', y='Engagement_Score', data=df)
plt.title('Engagement Score Distribution by Campaign Type')
plt.show()
```

```python
# 4. Bar: Average ROI by Company
plt.figure(figsize=(10,5))
roi_company = df.groupby('Company')['ROI'].mean().sort_values(ascending=False)
roi_company.plot(kind='bar', color='teal')
plt.title('Average ROI by Company')
plt.ylabel('Avg ROI')
plt.xticks(rotation=30)
plt.show()
```

```python
# 5. Heatmap: Engagement_Score vs Conversion_Rate
plt.figure(figsize=(6,4))
corr = df[['Engagement_Score','Conversion_Rate']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation: Engagement vs Conversion')
plt.show()
```

```python
# --- CUSTOMER SEGMENTATION ---


# 6. Count Plot: Target_Audience
plt.figure(figsize=(8,5))
sns.countplot(y='Target_Audience', data=df, order=df['Target_Audience'].value_counts().index)
plt.title('Distribution of Target Audience')
plt.show()
```

```python
# 7. Bar: Highest Conversion Rate by Customer_Segment for each Language
plt.figure(figsize=(12,5))
lang_seg = df.groupby(['Language','Customer_Segment'])['Conversion_Rate'].mean().reset_index()
sns.barplot(x='Language', y='Conversion_Rate', hue='Customer_Segment', data=lang_seg)
plt.title('Highest Conversion by Segment per Language')
plt.xticks(rotation=15)
plt.show()
```

```python
# 8. Box Plot: Acquisition_Cost across Customer_Segment by Channel_Used
plt.figure(figsize=(12,5))
sns.boxplot(x='Customer_Segment', y='Acquisition_Cost', hue='Channel_Used', data=df)
plt.title('Acquisition Cost by Segment and Channel')
plt.xticks(rotation=20)
plt.show()
```

```python
# 9. Bar: Avg Conversion_Rate by Language
plt.figure(figsize=(8,5))
df.groupby('Language')['Conversion_Rate'].mean().sort_values().plot(kind='barh', color='orange')
plt.title('Avg Conversion Rate by Language')
plt.show()
```

```python
# --- CHANNEL EFFECTIVENESS ---


# 10. Bar: Engagement_Score for Channels, segmented by Campaign_Type
plt.figure(figsize=(10,5))
eng = df.groupby(['Channel_Used','Campaign_Type'])['Engagement_Score'].mean().reset_index()
sns.barplot(x='Channel_Used', y='Engagement_Score', hue='Campaign_Type', data=eng)
plt.title('Engagement Score by Channel and Campaign Type')
plt.xticks(rotation=20)
plt.show()
```

```python
# 11. Pie: Total ROI across Channel_Used
roi_channel = df.groupby('Channel_Used')['ROI'].sum()
plt.figure(figsize=(7,7))
plt.pie(roi_channel, labels=roi_channel.index, autopct='%1.1f%%')
plt.title('Total ROI Distribution by Channel')
plt.show()
```

```python
# 12. Scatter: Clicks vs Impressions for each Channel_Used
plt.figure(figsize=(8,6))
sns.scatterplot(x='Impressions', y='Clicks', hue='Channel_Used', data=df, alpha=0.7)
plt.title('Clicks vs Impressions by Channel')
plt.show()
```

```python
# --- TIME-BASED ANALYSIS ---


# 13. Histogram: Duration
plt.figure(figsize=(7,5))
plt.hist(df['Duration'], bins=20, color='purple', edgecolor='black')
plt.title('Distribution of Campaign Duration')
plt.xlabel('Duration (days)')
plt.show()
```

```python
# 14. Line: Conversion_Rate over Date for each Company
plt.figure(figsize=(12,6))
for company in df['Company'].unique()[:5]: # top 5 for clarity
    temp = df[df['Company']==company]
    plt.plot(temp['Date'], temp['Conversion_Rate'], label=company, marker='o')
plt.legend()
plt.title('Conversion Rate Over Time by Company')
plt.xticks(rotation=30)
plt.show()
```

```python
# 15. Line: Engagement_Score trend over Date
plt.figure(figsize=(12,5))
df.groupby('Date')['Engagement_Score'].mean().plot()
plt.title('Engagement Score Trend Over Time')
plt.ylabel('Avg Engagement Score')
plt.show()
```

```python
# --- GEOGRAPHIC ANALYSIS ---


# 16. Bar: Which location has highest Acquisition_Cost
plt.figure(figsize=(8,5))
df.groupby('Location')['Acquisition_Cost'].mean().sort_values(ascending=False).plot(kind='bar', color='red')
plt.title('Highest Avg Acquisition Cost by Location')
plt.show()
```

```python
# 17. Bar: Conversion_Rate by Location categorized by Target_Audience
plt.figure(figsize=(12,6))
loc_aud = df.groupby(['Location','Target_Audience'])['Conversion_Rate'].mean().reset_index()
sns.barplot(x='Location', y='Conversion_Rate', hue='Target_Audience', data=loc_aud)
plt.title('Conversion Rate by Location and Audience')
plt.xticks(rotation=20)
plt.show()
```

```python
# 18. Pie: ROI by Location
roi_loc = df.groupby('Location')['ROI'].sum()
plt.figure(figsize=(7,7))
plt.pie(roi_loc, labels=roi_loc.index, autopct='%1.1f%%')
plt.title('ROI Proportion by Location')
plt.show()
```
